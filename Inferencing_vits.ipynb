```python
from transformers import VitsModel, AutoTokenizer
import torch

from transformers import AutoTokenizer, AutoModelForTextToWaveform

tokenizer = AutoTokenizer.from_pretrained("procit006/dutch_female_v1.0")
model = AutoModelForTextToWaveform.from_pretrained("procit006/dutch_female_v1.0")

text = "Ik heb uw postalcode genoteerd Is drie drie een een R S het juiste huisnummer?"
inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    output = model(**inputs).waveform
```

```python
import torch
from transformers import VitsTokenizer, VitsModel, set_seed

tokenizer = VitsTokenizer.from_pretrained("procit006/dutch_female_v1.0",cache_dir="./cache")
model = VitsModel.from_pretrained("procit006/dutch_female_v1.0",cache_dir="./cache")
```

```python
inputs = tokenizer(text="Ik heb uw postalcode genoteerd. Is drie drie een een R S het juiste huisnummer?", return_tensors="pt")
# inputs = int(inputs)
# inputs = tokenizer(phonetic_text, return_tensors="pt")

set_seed(555)  # make deterministic

with torch.no_grad():
   outputs = model(**inputs)

waveform = outputs.waveform[0]
```

```python
from IPython.display import Audio

Audio(waveform, rate=model.config.sampling_rate)
```
